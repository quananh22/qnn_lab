```python
# export_resnet50_to_onnx.py
import torch
import torchvision.models as models

# 1) Tải model pretrained ImageNet
model = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
model.eval()

# 2) Dummy input (NCHW) cho trace
dummy = torch.randn(1, 3, 224, 224)

# 3) Export ONNX (opset 13 là lựa chọn an toàn)
torch.onnx.export(
    model, dummy, "resnet50.onnx",
    input_names=["input"], output_names=["logits"],
    opset_version=13, do_constant_folding=True,
    dynamic_axes=None  # hoặc {"input":{0:"N"}, "logits":{0:"N"}} nếu cần batch linh hoạt
)
print("Saved resnet50.onnx")
```
